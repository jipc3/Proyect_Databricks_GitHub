## WIDGETS

```python
%python
dbutils.widgets.removeAll()
```

```python
dbutils.widgets.text("catalogName", "unit_catalog_explorer")
dbutils.widgets.text("groupName", "ext_users")
```

## CONSTANTS

```python
catalog =  dbutils.widgets.get("catalogName")
group =  dbutils.widgets.get("groupName")
```

## REVOKE CATALOG Y SCHEMA

```python
%sql
REVOKE ALL PRIVILEGES ON CATALOG ${catalogName} from `${groupName}`;
REVOKE CREATE, USE SCHEMA ON SCHEMA   ${catalogName}.uc_bronze from `${groupName}`;
REVOKE CREATE, USE SCHEMA ON SCHEMA  ${catalogName}.uc_silver from `${groupName}`;
REVOKE CREATE, USE SCHEMA ON SCHEMA  ${catalogName}.uc_golden from `${groupName}`;
```

## REVOKE EXTERNAL LOCATION

```python
%sql
REVOKE READ FILES, WRITE FILES ON EXTERNAL LOCATION `ext-bronze` from `${groupName}`;
REVOKE READ FILES, WRITE FILES ON EXTERNAL LOCATION `ext-silver` from `${groupName}`;
REVOKE READ FILES, WRITE FILES ON EXTERNAL LOCATION `ext-golden` from `${groupName}`;
```
