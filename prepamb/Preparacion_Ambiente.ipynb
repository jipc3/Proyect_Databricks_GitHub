```sql
%python
dbutils.widgets.removeAll()
```

```sql
create widget text storageName default "saccexplorer";
```

```sql
DROP CATALOG IF EXISTS unit_catalog_explorer CASCADE;
```

```sql
CREATE CATALOG IF NOT EXISTS unit_catalog_explorer;
```

```sql
CREATE SCHEMA IF NOT EXISTS unit_catalog_explorer.uc_raw;
CREATE SCHEMA IF NOT EXISTS unit_catalog_explorer.uc_bronze;
CREATE SCHEMA IF NOT EXISTS unit_catalog_explorer.uc_silver;
CREATE SCHEMA IF NOT EXISTS unit_catalog_explorer.uc_golden;
CREATE SCHEMA IF NOT EXISTS unit_catalog_explorer.uc_explorer;

CREATE VOLUME IF NOT EXISTS unit_catalog_explorer.uc_raw.datasets;
```

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS `ext-raw`
URL 'abfss://raw@${storageName}.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL acc_explorer_credential)
COMMENT 'Ubicación externa para las tablas raw del Data Lake';
```
