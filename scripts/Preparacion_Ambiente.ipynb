## WIDGETS

```sql
%python
dbutils.widgets.removeAll()
```

```sql
create widget text storageName default "saccexplorer";
create widget text catalogName default "unit_catalog_explorer";
```

## EXTERNAL LOCATION

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS `ext-raw`
URL 'abfss://raw@${storageName}.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL acc_explorer_credential)
COMMENT 'Ubicación externa para las tablas raw del Data Lake';
```

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS `ext-bronze`
URL 'abfss://bronze@${storageName}.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL acc_explorer_credential)
COMMENT 'Ubicación externa para las tablas bronze del Data Lake';
```

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS `ext-silver`
URL 'abfss://silver@${storageName}.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL acc_explorer_credential)
COMMENT 'Ubicación externa para las tablas silver del Data Lake';
```

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS `ext-golden`
URL 'abfss://golden@${storageName}.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL acc_explorer_credential)
COMMENT 'Ubicación externa para las tablas golden del Data Lake';
```

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS `ext-explorer`
URL 'abfss://explorer@${storageName}.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL acc_explorer_credential)
COMMENT 'Ubicación externa para los catalogo y schema explorer del Data Lake';
```

## CATALOG Y SCHEMA

```sql
DROP CATALOG IF EXISTS ${catalogName} CASCADE;
```

```sql
CREATE CATALOG IF NOT EXISTS ${catalogName}
MANAGED LOCATION 'abfss://explorer@${storageName}.dfs.core.windows.net/'
```

```sql
CREATE SCHEMA IF NOT EXISTS ${catalogName}.uc_raw;
CREATE SCHEMA IF NOT EXISTS ${catalogName}.uc_bronze;
CREATE SCHEMA IF NOT EXISTS ${catalogName}.uc_silver;
CREATE SCHEMA IF NOT EXISTS ${catalogName}.uc_golden;
```
