## WIDGETS

```python
%python
dbutils.widgets.removeAll()
```

```python
dbutils.widgets.text("catalogName", "unit_catalog_explorer")
dbutils.widgets.text("groupName", "ext_users")
```

## CONSTANTS

```python
catalog =  dbutils.widgets.get("catalogName")
group =  dbutils.widgets.get("groupName")
```

## GRANTS CATALOG Y SCHEMA

```python
%sql
GRANT ALL PRIVILEGES ON CATALOG ${catalogName} TO `${groupName}`;
GRANT CREATE, USE SCHEMA ON SCHEMA  ${catalogName}.uc_bronze to `${groupName}`;
GRANT CREATE, USE SCHEMA ON SCHEMA  ${catalogName}.uc_silver to `${groupName}`;
GRANT CREATE, USE SCHEMA ON SCHEMA  ${catalogName}.uc_golden to `${groupName}`;
```

```python
%sql

SHOW GRANTS ON CATALOG ${catalogName};
SHOW GRANTS ON SCHEMA ${catalogName}.uc_bronze;
SHOW GRANTS ON SCHEMA ${catalogName}.uc_silver;
SHOW GRANTS ON SCHEMA ${catalogName}.uc_golden;
```

## GRANTS EXTERNAL LOCATION

```python
%sql
GRANT READ FILES, WRITE FILES ON EXTERNAL LOCATION `ext-bronze` to `${groupName}`;
GRANT READ FILES, WRITE FILES ON EXTERNAL LOCATION `ext-silver` to `${groupName}`;
GRANT READ FILES, WRITE FILES ON EXTERNAL LOCATION `ext-golden` to `${groupName}`;
```

```python
%sql

SHOW GRANTS ON EXTERNAL LOCATION `ext-bronze`;
SHOW GRANTS ON EXTERNAL LOCATION `ext-silver`;
SHOW GRANTS ON EXTERNAL LOCATION `ext-golden`;
```
